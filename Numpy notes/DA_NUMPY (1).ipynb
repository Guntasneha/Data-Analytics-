```python
#choice with probability
import numpy as np
arr=[10,20,30]
result=np.random.choice(
    arr,
    size=3,
    p=[0.2,0.5,0.3]
)
print(result)
```

```python
# numpy aggregations in single array
import numpy as np
arr=np.array([10,20,30,40,50])
print(np.sum(arr))
print(np.min(arr))
print(np.max(arr))
print(np.mean(arr))
print(np.var(arr))
print(np.std(arr))
print(np.median(arr))
```

```python
#percentile ---> 0 to 100
print(np.percentile(arr,100)) #q4
print(np.percentile(arr,50)) #q3
print(np.percentile(arr,75)) #q2
print(np.percentile(arr,25)) #q1
```

```python
#quantile --> 0 to 1
print(np.quantile(arr,1.0))
print(np.quantile(arr,0.75))
print(np.quantile(arr,0.50))
print(np.quantile(arr,0.25))
```

```python
# argmax---> it returns index
print(np.argmax(arr))
print(np.argmin(arr))
```
